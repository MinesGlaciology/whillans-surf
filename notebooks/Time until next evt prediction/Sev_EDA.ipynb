```python
# Sam Brown
# sam_brown@mines.edu
# July 28
# Goal, look more into the pattern of slip severity and time until next event.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.read_csv("/Users/sambrown04/Documents/SURF/Preproc_data/10-18.csv")
```

```python
df['time_until'] = df['time_since'].shift(-1)
```

```python
df.head()
```

```python
fig, ax = plt.subplots()
ax.scatter(df['sev_stds'], df['time_until'])
ax.set_xlim([-5,5])
ax.set_ylim([0,2200])
ax.set_xlabel('Slip Severity')
ax.set_ylabel('Time Until Next Event')
ax.grid(True)
```

```python
fig, ax = plt.subplots()
ax.scatter(df['sev_stds'], (df['tide_deriv']))
ax.set_xlim([-2,5])
ax.set_xlabel('Slip Severity')
ax.set_ylabel('Tide_deriv')
ax.grid(True)
```

```python
fig, ax = plt.subplots()
ax.scatter(df['sev_stds'], df['slip_size_standardized'])
ax.set_xlim([-3,5])
ax.set_xlabel('Slip Severity')
ax.set_ylabel('Slip Size')
```
